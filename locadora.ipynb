```python


portifolio = [
    ['Chevrolet Tracker',120],
    ['Chevrolet Onix',90],
    ['Chevrolet Spin',150],
    ['Hyundai HB20',85], 
    ['Hyundai Tucson',120],
    ['Fiat Uno',60],
    ['Fiat Mobi',70],
    ['Fiat Pulse',130]
]

carros_alugados = []




while True:

    print('''=========
    Bem vindo à locadora de carros!
    ========''')
    print('''O que deseja fazer?
    0 - Mostrar portifólio | 1 - Alugar um carro | 2 - Devolver um carro''')
    escolha = int(input())
    
    while True:
        if escolha not in [0,1,2]:
            print('Opção invalida:')
            escolha = int(input('Escolha uma opção valida: '))
        else:
            break
        
    if escolha == 0:
        for i in portifolio:
            print(f'[{portifolio.index(i)}] {i[0]} - R$ {i[1]} / dia')
    
        print('''\n\n==========
    0 - CONTINUAR | 1 - SAIR''')
        confirmar = int(input())
    
        while True:
          if confirmar not in [0,1]:
            print('Opção invalida:')
            confirmar = int(input('Escolha uma opção valida: '))
          else:
                break
    
        if confirmar == 0:
            continue
        else: 
            break
              
    
    elif escolha == 1:
        print('[ALUGAR] Dê uma olhada em nosso portifólio.\n')
        
        for i in portifolio:
            print(f'[{portifolio.index(i)}] {i[0]} - R$ {i[1]} / dia')
    
        print('''==========
    Escolha o código do carro:''')
        codigo = int(input())
        while True:
            if codigo not in list(range(0,len(portifolio))):
                print('Opção invalida:')
                codigo = int(input('Escolha uma opção valida: '))
            else:
                break    
        
        print('Por quantos dias deseja alugar o carro?')
        dias = int(input())
        
        print(f'Você escolheu {portifolio[codigo][0]} por {dias} dias.')
        print(f'O aluguel totalizaria R$ {portifolio[codigo][1] * dias}. Deseja alugar?')
        print('\n 0 - SIM | 1 - NÃO\n')
    
        confirmar = int(input())
        
        while True:
            if confirmar not in [0,1]:
                print('Opção invalida:')
                confirmar = int(input('Escolha uma opção valida: '))
            else:
                break
                
        if confirmar == 0:
            print(f'Parabens você alugou o {portifolio[codigo][0]} por {dias} dias')
            registro_carro = [codigo,portifolio[codigo]]
            portifolio.pop(codigo)
            carros_alugados.append(registro_carro)
            continue
        else:
            continue

    elif escolha == 2:
        carros_a_devolver = len(carros_alugados)
        if not carros_a_devolver:
            print('Você não tem carros a devolver!\n\n\n\n')
            print('''\n\n==========
    0 - CONTINUAR | 1 - SAIR''')
            confirmar = int(input())
            while True:
              if confirmar not in [0,1]:
                print('Opção invalida:')
                confirmar = int(input('Escolha uma opção valida: '))
              else:
                    break
        
            if confirmar == 0:
                continue
            else: 
                break

        else:
            print('Segue a lista de carros alugados. Qual você deseja devolver?')
            for indice,valor in enumerate(carros_alugados):
                print(f'[{indice}] {valor[1][0]} - R$ {valor[1][1]}')
            print('\n\n Escolha o código do carro que deseja devolver:')
            codigo_devolucao = int(input())

            while True:
                if codigo_devolucao not in list(range(0,carros_a_devolver)):
                    print('Opção invalida:')
                    codigo_devolucao = int(input('Escolha uma opção valida: '))
                else: 
                    break

            print(f'Você quer devolver o {carros_alugados[codigo_devolucao][1][0]}?')
            print('0 - SIM | 1 - NÃO')
            confirmar = int(input())

            while True:
                if confirmar not in [0,1]:
                    print('Opção invalida:')
                    confirmar = int(input('Escolha uma opção valida: '))
                else:
                    break

            if confirmar == 0:
                print(f'Obrigado por devolver o carro {carros_alugados[codigo_devolucao][1][0]}')
                portifolio.insert(carros_alugados[codigo_devolucao][0],carros_alugados[codigo_devolucao][1])
                carros_alugados.pop(codigo_devolucao)
                continue

            else:
                 continue   
            
                    
                    
        
```
